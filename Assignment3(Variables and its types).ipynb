```python
def demo():
    #Variable and Built in Data types
    integer=20 #integer
    flt=25.5 #float
    string="Python" #String
    boolean= True #Boolean
    lst=[10,20,30,40,50] #List
    tpl=(45,67,89,90) #Tuple
    dictnry={'Id':1,'Name':"Karan"} #dictionary 
#Print Variables and their data types
    print("Variables and their data types")
    print(f"Integer : {integer} (Type :{type(integer)})")
    print(f"Float : {flt} (Type :{type(flt)})")
    print(f"String : {string} (Type :{type(string)})")
    print(f"Boolean : {boolean} (Type :{type(boolean)})")
    print(f"List: {lst} (Type :{type(lst)})")
    print(f"Tuple : {tpl} (Type :{type(tpl)})")
    print(f"Dictionary : {dictnry} (Type :{type(dictnry)})")
#Arithmetic Operators
    print("\nArithmetic Operators")
    print(f"{integer}+{flt}={integer+flt}")
    print(f"{integer}-{flt}={integer-flt}")
    print(f"{integer}*{flt}={integer*flt}")
    print(f"{integer}/{flt}={integer/flt}")
    print(f"{integer}%3={integer%3}")
    print(f"{integer}**2={integer**2}")
    print(f"{integer}//3={integer//3}")
#Comparison Operators
    print("\nComparison Operators")
    print(f"{integer}==10:{integer==10}") 
    print(f"{integer}!=10:{integer!=10}") 
    print(f"{integer}>5:{integer>5}") 
    print(f"{integer}<15:{integer<15}") 
    print(f"{integer}>=10:{integer>=10}") 
    print(f"{integer}<=10:{integer<=10}") 
#Logical Operators
    print("\nLogical Operators")
    print(f"True and False : {True and False}")
    print(f"True or False : {True or False}")
    print(f"not True  : {not True}")
#Assignment Operators
    print("\nAssignment Operators")
    x=5
    print(f"x=5:{x}")
    x+=3
    print(f"x+=3:{x}")
    x-=2
    print(f"x-=2:{x}")
    x*=2
    print(f"x*=2:{x}")
    x/=3
    print(f"x/=3:{x}")
    x%=2
    print(f"x%=2:{x}")
    x**=3
    print(f"x**=3:{x}")
    x//=2
    print(f"x//=2:{x}")
if __name__=="__main__":
        demo()
```
