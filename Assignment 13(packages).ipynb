```python
%%writefile my_package1/math_operations.py
#math_operations.py

def add(a,b):
    return a+b

def substract(a,b):
    return a-b

def multiply(a,b):
    return a*b

def divide(a,b):
    if b==0:
        raise ValueError("Cnnot divide by Zero ")
    return a/b
```

```python
#main.py
from my_package1 import math_operations

#Using the functions from math_operation module
a=20
b=10

print(f"{a} + {b} = {math_operations.add(a,b)}")
print(f"{a} - {b} = {math_operations.substract(a,b)}")
print(f"{a} * {b} = {math_operations.multiply(a,b)}")
print(f"{a} / {b} = {math_operations.divide(a,b)}")
```
