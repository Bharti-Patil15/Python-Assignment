```python
def main():
    num=input("Entered numbers seperated by spaces : ")
    numlst=list((map(int ,num.split())))
    ttl_sum=sum(numlst)
    max_value=max(numlst)
    min_value=min(numlst)
    srt_lst=sorted(numlst)
    
    print(f"Sum = {ttl_sum}")
    print(f"Max = {max_value}")
    print(f"Min = {min_value}")
    print(f"Sorted = {srt_lst}")
    
if __name__=="__main__":
    main()
```
