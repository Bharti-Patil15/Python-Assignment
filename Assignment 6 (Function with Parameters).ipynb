```python
def add_num(a,b):
    """Calculate the sum"""
    return a+b
def mult_num(a,b):
     """Calculate the multiplication""" 
     return a*b    
def main():
    n1=float(input("Enter the 1st Number"))
    n2=float(input("Enter the 2nd Number"))

    sum_result=add_num(n1,n2)
    prdct_result=mult_num(n1,n2)

    print(f"The Sum of Entered number is :{sum_result}")
    print(f"The Product of Entered number is :{prdct_result}")

if __name__=="__main__":
    main()
    
```
