```python
def con_age(names,ages):
    age1=list(map(lambda age: age * 12,ages))
    name_dic={name:age for name ,age in zip(names,age1)}
    return name_dic
def main():
    names=['abc' ,'pqr','xyz']
    ages=[25,30,40]
    name_dic=con_age(names,ages)
    print("Name to age in Months Conversion")
    for name,age_mon in name_dic.items():
        print(f"{name} : {age_mon} months")
if __name__ == "__main__":
    main()
```
