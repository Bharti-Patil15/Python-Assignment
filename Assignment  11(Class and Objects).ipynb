```python
class A:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def demo(self):
        print("Name :",self.name)
        print("Age :",self.age)
        
    def temp(self):
        print("Hello ",self.name)
        print("My age is ",self.age)

o1=A("Alex",21)
o2=A("John",25)

o1.demo()
print()

o2.demo()
print()

o1.temp()
o2.temp()


        
```
